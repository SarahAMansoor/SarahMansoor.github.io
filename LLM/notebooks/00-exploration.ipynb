This notebook is intended for exploratory data analysis and experimentation with the RAG (Retrieval-Augmented Generation) model.

## Objectives
- Load and preprocess the PDF document.
- Explore the text extraction and formatting capabilities.
- Analyze the statistics of the extracted text.
- Experiment with the retriever model to fetch relevant information.

```python
# Import necessary libraries
import pandas as pd
from LLM.RAG.rag import open_and_read_pdf

# Load the PDF document
pdf_path = '../data/human-nutrition-text.pdf'
pages_and_texts = open_and_read_pdf(pdf_path)

# Convert the extracted data into a DataFrame for analysis
df = pd.DataFrame(pages_and_texts)
df.head()
```

```python
# Get basic statistics of the extracted text
df.describe().round(2)
```

## Text Analysis

In this section, we will analyze the text extracted from the PDF to understand its structure and content.

```python
# Sample some pages to inspect the text
import random

sample_pages = random.sample(pages_and_texts, k=3)
for page in sample_pages:
    print(f"Page {page['page_number']}:")
    print(page['text'][:500])  # Print the first 500 characters of the page text
    print("\n---\n")
```
